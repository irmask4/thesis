## Coefficient threshold 0.01

```python
import pandas as pd
import os

# Define folder path for screening size 10 at threshold 0.1
base_path = "/Users/irmaskaljo/codedev/masters_thesis/Data/GRN/Yeast/scr_size_10_r5/0.01"
module_names = [f"{i}" for i in range(1, 8)]  # 7 local modules

# Load and label each file
dfs_with_names = []
for i, name in enumerate(module_names, start=1):
    filename = f"r5_fpkm_6400_net_m_{i}_0.01_list"
    filepath = os.path.join(base_path, filename)
    df = pd.read_csv(filepath, sep="\t", names=["Reg", "Weight"], header=0)
    df["Module"] = name
    dfs_with_names.append(df)

# Merge all modules
consolidated_df = pd.concat(dfs_with_names, ignore_index=True)

# Split interaction column into 'Regulator' and 'TargetGene'
consolidated_df[["Regulator", "TargetGene"]] = consolidated_df["Reg"].str.split("_", expand=True)
# Reorder columns for clarity
consolidated_df = consolidated_df[[
    "Reg", "Regulator", "TargetGene", "Weight" ,"Module"]]

# Save to CSV
consolidated_df.to_csv("Yeast_consolidated_scr10_r5_t0.01.csv", index=False)

# Preview
print(consolidated_df.head())
```

```python
# print the number of gene interactions across all modules
total_interactions = consolidated_df.shape[0]
print(f"Total interactions across all modules: {total_interactions}")
```

## Coefficient threshold 0.05

```python
import pandas as pd
import os

# Define folder path for screening size 10 at threshold 0.05 (update version)
base_path = "/Users/irmaskaljo/codedev/masters_thesis/Data/GRN/Yeast/scr_size_10_r5/0.05_update"
module_names = [f"{i}" for i in range(1, 8)]  # Modules 1 to 7

# Load and label each file
dfs_with_names = []
for i, name in enumerate(module_names, start=1):
    filename = f"r5_fpkm_6400_net_m_{i}_0.05_list.update"
    filepath = os.path.join(base_path, filename)
    
    # Load file with biotype columns
    df = pd.read_csv(filepath, sep="\t", names=["Reg", "Weight", "RegulatorBiotype", "TargetBiotype"], header=0)
    df["Module"] = name
    
    # Split interaction column
    df[["Regulator", "TargetGene"]] = df["Reg"].str.split("_", expand=True)
    
    dfs_with_names.append(df)

# Merge all modules
consolidated_df = pd.concat(dfs_with_names, ignore_index=True)

# Reorder columns for clarity
consolidated_df = consolidated_df[[
    "Reg", "Regulator", "TargetGene", "Weight" ,"Module", "RegulatorBiotype", "TargetBiotype"
]]

# Save to CSV
consolidated_df.to_csv("Yeast_consolidated_scr10_r5_t0.05_biotype.csv", index=False)

# Preview
print(consolidated_df.head())
```
