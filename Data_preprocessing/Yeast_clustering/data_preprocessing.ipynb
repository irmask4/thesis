```python
# Base path (before the module number)
base_path = "/Users/irmaskaljo/codedev/thesis/Data/Cluster_F10/Genexpression/cluster_gene_expression_based_on_r5_fpkm_6400_net_"

# Collect all dataframes
dfs = []

for i in range(1, 8):  # for modules 1 through 7
    file_path = f"{base_path}m_{i}_0.01"
    df = pd.read_csv(file_path, sep="\t")
    df["Module"] = f"{i}"  # add module column
    dfs.append(df)

# Concatenate all into one dataframe
consolidated_df = pd.concat(dfs, ignore_index=True)
consolidated_df = consolidated_df.rename(columns={"Unnamed: 0": "GeneName"})


consolidated_df.head()
```
