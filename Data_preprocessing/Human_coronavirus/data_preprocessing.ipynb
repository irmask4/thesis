```python
import pandas as pd

l1 = pd.read_csv("/Users/irmaskaljo/codedev/thesis/Data/GRN/Human_coronavirus/R5T0.5/L1_r5_t0.5_net_entrez", sep="\t", names = ["Reg", "Weight"], header=0)
l2 = pd.read_csv("/Users/irmaskaljo/codedev/thesis/Data/GRN/Human_coronavirus/R5T0.5/L2_r5_t0.5_net_entrez", sep="\t", names = ["Reg", "Weight"], header=0)
l3 = pd.read_csv("/Users/irmaskaljo/codedev/thesis/Data/GRN/Human_coronavirus/R5T0.5/L3_r5_t0.5_net_entrez", sep="\t", names = ["Reg", "Weight"],  header=0)
l4 = pd.read_csv("/Users/irmaskaljo/codedev/thesis/Data/GRN/Human_coronavirus/R5T0.5/L4_r5_t0.5_net_entrez", sep="\t", names = ["Reg", "Weight"],  header=0)
l5 = pd.read_csv("/Users/irmaskaljo/codedev/thesis/Data/GRN/Human_coronavirus/R5T0.5/L5_r5_t0.5_net_entrez", sep="\t", names = ["Reg", "Weight"], header=0)
```

```python
# make a condolidated df with all dataframes and a source column
#add source column to each dataframe
def add_source_column(df, source_name):
    df = df.copy()
    df['Module'] = source_name
    return df

# List of dataframes with their names
dfs_with_names = [
    (l1, 'L1'), (l2, 'L2'), (l3, 'L3'), (l4, 'L4'), (l5, 'L5')
]
#consolidate all dataframes into one and add source
df_list = [add_source_column(df, name) for df, name in dfs_with_names]
consolidated_df = pd.concat(df_list, ignore_index=True)

consolidated_df.head()
```

```python
# Split the 'Reg' column into 'Regulator' and 'TargetGene'
consolidated_df[['Regulator', 'TargetGene']] = consolidated_df['Reg'].str.split('_', expand=True)
consolidated_df.head()
```

```python
consolidated_df.to_csv("HS_consolidated_r5_t0.5.csv", index=False)
```
