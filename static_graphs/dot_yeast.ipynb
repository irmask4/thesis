```python
import pandas as pd
from graphviz import Digraph
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.colors as mcolors

# Load the data
df = pd.read_csv('/Users/irmaskaljo/codedev/thesis/Data_preprocessing/Yeast/scr_size_10/Yeast_consolidated_scr10_r5_t0.1_biotype.csv')

# Get unique modules
unique_modules = sorted(df['Module'].unique())

# Create reversed Viridis color map
cmap = cm.get_cmap('viridis_r', len(unique_modules))  # 'viridis_r' is reversed Viridis

# Normalize module indices to range 0-1
norm = mcolors.Normalize(vmin=0, vmax=len(unique_modules) - 1)

# Generate color mapping for each module
module_colors = {str(mod): mcolors.to_hex(cmap(i)) for i, mod in enumerate(unique_modules)}

# Create a directed graph
dot = Digraph(format='png', engine='dot')  # 'dot' for hierarchical layout

# Set graph attributes
dot.attr(overlap='false')
dot.attr(splines='true')
dot.attr(dpi='300')
dot.attr(fontsize='12')
dot.attr(rankdir='LR')

# Add nodes
nodes = set(df['Regulator']).union(set(df['TargetGene']))
for node in nodes:
    dot.node(node, shape='box', style='filled', fillcolor='white')

# Create a dictionary to store modules for each edge
edge_modules = {}

# Add edges with modules (no coefficients)
for _, row in df.iterrows():
    regulator = row['Regulator']
    target = row['TargetGene']
    module = row['Module']

    edge_key = (regulator, target)
    if edge_key not in edge_modules:
        edge_modules[edge_key] = set()
    edge_modules[edge_key].add(str(module))

# Add the edges with combined module names (no coefficients)
for (regulator, target), modules in edge_modules.items():
    module_list = sorted(modules)
    label = ', '.join(module_list)

    # Choose color based on the first module
    color = module_colors.get(module_list[0], 'gray')

    # Draw the edge with the combined label (module names only)
    dot.edge(regulator, target, color=color, label=label, fontsize='10')

# Save and render the graph
dot.render('gene_network_graph_multi_module_combined_0.1', view=True)
```

```python
import pandas as pd
from graphviz import Digraph
from collections import defaultdict

# === Load Data ===
df = pd.read_csv('/Users/irmaskaljo/codedev/thesis/Data_preprocessing/Yeast/scr_size_10/Yeast_consolidated_scr10_r5_t0.1_biotype.csv')

# === Module Colors ===
import matplotlib as mpl
unique_modules = sorted(df['Module'].unique())
cmap = mpl.colormaps.get_cmap('viridis_r')
colors = [cmap(i / (len(unique_modules)-1)) for i in range(len(unique_modules))]
module_colors = {str(mod): mpl.colors.to_hex(colors[i]) for i, mod in enumerate(unique_modules)}

# === Initialize Graphviz ===
dot = Digraph(format='png', engine='dot')
dot.attr(overlap='false', splines='true', dpi='300', fontsize='12', rankdir='LR')

# === Collect Nodes & Edge Info ===
nodes = set(df['Regulator']).union(set(df['TargetGene']))
edge_modules = defaultdict(set)
out_degree = defaultdict(int)
node_biotypes = {}

for _, row in df.iterrows():
    regulator = row['Regulator']
    target = row['TargetGene']
    module = str(row['Module'])
    biotype = row['TargetBiotype']

    edge_modules[(regulator, target)].add(module)
    out_degree[regulator] += 1
    node_biotypes[target] = biotype

# === Determine Terminal Nodes ===
terminal_nodes = [n for n in set(df['TargetGene']) if out_degree[n] == 0]

# === Group Candidates for Supernodes ===
group_candidates = defaultdict(list)
for (reg, target), modules in edge_modules.items():
    if target in terminal_nodes:
        modules_tuple = tuple(sorted(modules, key=lambda x: int(x[1:]) if x[1:].isdigit() else x))
        biotype = node_biotypes[target]
        group_candidates[(reg, modules_tuple, biotype)].append(target)

# === Filter True Supernodes ===
valid_supernodes = {k: v for k, v in group_candidates.items() if len(v) > 1}

# === Track Nodes in Supernodes ===
nodes_in_supernodes = set()
for targets in valid_supernodes.values():
    nodes_in_supernodes.update(targets)

# === Add Non-supernode Nodes ===
for node in nodes:
    if node not in nodes_in_supernodes:
        dot.node(node, shape='box', style='filled', fillcolor='white')

# === Render Supernodes ===
for (reg, modules, biotype), targets in valid_supernodes.items():
    sn_name = f'group_{reg}_{"_".join(modules)}_{biotype}'
    label = '\\n'.join(targets)
    dot.node(sn_name, shape='box', style='filled', fillcolor='white', label=label)

    # Edge from regulator to supernode
    color = 'gray' if len(modules) > 1 else module_colors.get(modules[0], 'gray')
    dot.edge(reg, sn_name, label=', '.join(modules), color=color, fontsize='10')

# === Add Remaining Edges ===
for (reg, target), modules in edge_modules.items():
    if target in nodes_in_supernodes:
        continue
    module_list = sorted(modules)
    label = ', '.join(module_list)
    color = 'gray' if len(modules) > 1 else module_colors.get(module_list[0], 'gray')
    dot.edge(reg, target, label=label, color=color, fontsize='10')

# === Save & Render ===
dot.render('gene_network_graph_multi_module_supernodes_scr10_0.1', view=True)
```

```python
import pandas as pd
from graphviz import Digraph
from collections import defaultdict
import matplotlib as mpl
from matplotlib.colors import to_rgb, to_hex

# === Load Data ===
df = pd.read_csv('/Users/irmaskaljo/codedev/thesis/Data_preprocessing/Yeast/scr_size_10/Yeast_consolidated_scr10_r5_t0.1_biotype.csv')

# === Module Colors ===
unique_modules = sorted(df['Module'].unique())
cmap = mpl.colormaps.get_cmap('viridis_r')
colors = [cmap(i / (len(unique_modules)-1)) for i in range(len(unique_modules))]
module_colors = {str(mod): to_hex(colors[i]) for i, mod in enumerate(unique_modules)}

# === Helper to Darken Colors ===
def darken_color(hex_color, factor=0.7):
    rgb = to_rgb(hex_color)
    dark_rgb = tuple(max(0, c*factor) for c in rgb)
    return to_hex(dark_rgb)

# === Initialize Graphviz ===
dot = Digraph(format='png', engine='dot')
dot.attr(
    overlap='false',
    splines='true',
    dpi='300',
    fontsize='16',
    rankdir='LR',
    ranksep='0.1',  # reduced vertical separation
    nodesep='0.15'  # reduced horizontal separation
)

# === Collect Nodes & Edge Info ===
nodes = set(df['Regulator']).union(set(df['TargetGene']))
edge_modules = defaultdict(set)
out_degree = defaultdict(int)
node_biotypes = {}

for _, row in df.iterrows():
    regulator = row['Regulator']
    target = row['TargetGene']
    module = str(row['Module'])
    biotype = row['TargetBiotype']

    edge_modules[(regulator, target)].add(module)
    out_degree[regulator] += 1
    node_biotypes[target] = biotype

# === Determine Terminal Nodes ===
terminal_nodes = [n for n in set(df['TargetGene']) if out_degree[n] == 0]

# === Group Candidates for Supernodes ===
group_candidates = defaultdict(list)
for (reg, target), modules in edge_modules.items():
    if target in terminal_nodes:
        modules_tuple = tuple(sorted(modules, key=lambda x: int(x[1:]) if x[1:].isdigit() else x))
        biotype = node_biotypes[target]
        group_candidates[(reg, modules_tuple, biotype)].append(target)

# === Filter True Supernodes ===
valid_supernodes = {k: v for k, v in group_candidates.items() if len(v) > 1}

# === Track Nodes in Supernodes ===
nodes_in_supernodes = set()
for targets in valid_supernodes.values():
    nodes_in_supernodes.update(targets)

# === Add Non-supernode Nodes ===
for node in nodes:
    if node not in nodes_in_supernodes:
        dot.node(node, shape='box', style='filled', fillcolor='white')

# === Render Supernodes ===
for (reg, modules, biotype), targets in valid_supernodes.items():
    sn_name = f'group_{reg}_{"_".join(modules)}_{biotype}'
    label = '\\n'.join(targets)
    dot.node(sn_name, shape='box', style='filled', fillcolor='white', label=label)

    color = 'gray' if len(modules) > 1 else module_colors.get(modules[0], 'gray')
    label_color = darken_color(color, factor=0.6)

    dot.edge(
        reg, sn_name,
        label=', '.join(modules),
        color=color,
        fontsize='12',
        labelfontcolor=label_color,
        labeldistance='0.1',
        labelangle='0'
    )

# === Add Remaining Edges ===
for (reg, target), modules in edge_modules.items():
    if target in nodes_in_supernodes:
        continue
    module_list = sorted(modules)
    label = ', '.join(module_list)
    color = 'gray' if len(modules) > 1 else module_colors.get(module_list[0], 'gray')
    label_color = darken_color(color, factor=0.6)
    
    dot.edge(
        reg, target,
        label=label,
        color=color,
        fontsize='12',
        labelfontcolor=label_color,
        labeldistance='2',
        labelangle='0'
    )

# === Save & Render ===
dot.render('gene_network_graph_scr10_0.1_edgerep', view=True)
```

```python
# Collect edges with module labels for D3.js visualization
import pandas as pd
import json

df = pd.read_csv(
    '/Users/irmaskaljo/codedev/thesis/Data_preprocessing/Yeast/scr_size_10/Yeast_consolidated_scr10_r5_t0.1_biotype.csv'
)

# Collect edges with module labels
edges = []
for _, row in df.iterrows():
    regulator = row['Regulator']
    target = row['TargetGene']
    module = str(row['Module'])
    edges.append({"source": regulator, "target": target, "module": module})

# Unique nodes
nodes = sorted(set(df['Regulator']).union(set(df['TargetGene'])))

# Save JSON for D3
data = {"nodes": [{"id": n} for n in nodes], "links": edges}
with open("graph.json", "w") as f:
    json.dump(data, f, indent=2)
```
