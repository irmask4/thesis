```python
import pandas as pd
from graphviz import Digraph

# Load the data
df = pd.read_csv('/Users/irmaskaljo/codedev/thesis/Data_preprocessing/Human_coronavirus/HS_consolidated_r5_t0.125.csv')

# Define color map for modules
module_colors = {
    'L1': 'mediumseagreen',
    'L2': 'gold',
    'L3': 'sienna',
    'L4': 'mediumpurple',
    'L5': 'black'
}

# Create a directed graph
dot = Digraph(format='png', engine='dot')  # 'dot' for hierarchical layout

# Set graph attributes
dot.attr(overlap='false')
dot.attr(splines='true')
dot.attr(dpi='300')
dot.attr(fontsize='12')
dot.attr(rankdir='LR')

# Add nodes
nodes = set(df['Regulator']).union(set(df['TargetGene']))
for node in nodes:
    dot.node(node, shape='box', style='filled', fillcolor='white')

# Create a dictionary to store modules and coefficients for each edge
edge_modules = {}

# Add edges with modules and coefficients
for _, row in df.iterrows():
    regulator = row['Regulator']
    target = row['TargetGene']
    module = row['Module']
    coeff = row['Weight']

    # Combine modules and coefficients for the same gene interaction (edge)
    edge_key = (regulator, target)
    if edge_key not in edge_modules:
        edge_modules[edge_key] = []
    edge_modules[edge_key].append((module, coeff))

# Add the edges with sorted module names and coefficients
for (regulator, target), modules in edge_modules.items():
    # Get unique modules for this edge and sort them by module name
    unique_modules = list(set(modules))
    sorted_modules = sorted(unique_modules, key=lambda x: int(x[0][1:]))  # Sort by number after 'L'
    
    labels = ', '.join([f'{module[0]}({module[1]:.2f})' for module in sorted_modules])
    
    # Set edge color to gray if there are multiple modules, else default from color map
    color = 'gray' if len(sorted_modules) > 1 else module_colors.get(sorted_modules[0][0], 'gray')
    
    # Draw the edge with sorted labels and appropriate color
    dot.edge(regulator, target, color=color, label=labels, fontsize='10')

# Save and render the graph
dot.render('gene_network_graph_multi_module_combined', view=True)
```

```python
# Final graph
import pandas as pd
from graphviz import Digraph
from collections import defaultdict, Counter

# === Load Data ===
df = pd.read_csv('/Users/irmaskaljo/codedev/thesis/Data_preprocessing/Human_coronavirus/HS_consolidated_r5_t0.25.csv')

# === Module Color Map ===
module_colors = {
    'L1': 'mediumseagreen',
    'L2': 'gold',
    'L3': 'sienna',
    'L4': 'mediumpurple',
    'L5': 'black'
}

# === Initialize Graphviz Digraph ===
dot = Digraph(format='png', engine='dot')
dot.attr(overlap='false', splines='true', dpi='300', fontsize='12', rankdir='LR')

# === Collect Nodes & Edge Info ===
nodes = set(df['Regulator']).union(set(df['TargetGene']))
edge_modules = defaultdict(list)
out_degree = defaultdict(int)
node_modules = defaultdict(list)

for _, row in df.iterrows():
    regulator = row['Regulator']
    target = row['TargetGene']
    module = row['Module']
    coeff = row['Weight']

    edge_modules[(regulator, target)].append((module, coeff))
    out_degree[regulator] += 1
    node_modules[target].append(module)

# === Determine Terminal Nodes ===
terminal_nodes = [n for n in set(df['TargetGene']) if out_degree[n] == 0]

# === Group Potential Supernodes: regulator → targets with exact module set
group_candidates = defaultdict(list)
for (reg, target), mod_list in edge_modules.items():
    if target in terminal_nodes:
        modules = tuple(sorted(set(m[0] for m in mod_list), key=lambda x: int(x[1:])))
        group_candidates[(reg, modules)].append((target, mod_list))

# === Filter True Supernodes: All targets must be regulated by same module set
valid_supernodes = {}
for (reg, modules), targets in group_candidates.items():
    if len(targets) > 1:
        all_match = all(
            tuple(sorted(set(m[0] for m in mod_list), key=lambda x: int(x[1:]))) == modules
            for _, mod_list in targets
        )
        if all_match:
            valid_supernodes[(reg, modules)] = targets

# === Identify Terminal Genes Included in Supernodes
nodes_in_supernodes = set()
for entries in valid_supernodes.values():
    for target, _ in entries:
        nodes_in_supernodes.add(target)

# === Add Non-supernode Nodes ===
for node in nodes:
    if node not in nodes_in_supernodes:
        dot.node(node, shape='box', style='filled', fillcolor='white')

# === Track Added Edges ===
added_edges = set()

# === Render Supernodes ===
for (reg, modules), entries in valid_supernodes.items():
    sn_name = f'group_{reg}_{"_".join(modules)}'

    # Build supernode label
    label_lines = []
    for gene, mod_list in entries:
        label_parts = [f'{mod}, {coeff:.2f}' for mod, coeff in mod_list if mod in modules]
        line = f'{gene}, ' + ', '.join(label_parts)
        label_lines.append(line)
        added_edges.add((reg, gene))

    wrapped_label = '\\n'.join(sorted(label_lines))
    dot.node(sn_name, shape='box', style='filled', fillcolor='white', label=wrapped_label)

    # Edge from regulator to supernode
    label = ', '.join(modules)
    color = 'gray' if len(modules) > 1 else module_colors.get(modules[0], 'gray')
    dot.edge(reg, sn_name, label=label, color=color, fontsize='13')

# === Handle Remaining Edges ===
for (reg, target), mod_list in edge_modules.items():
    if (reg, target) in added_edges or target in nodes_in_supernodes:
        continue
    modules = tuple(sorted(set(m[0] for m in mod_list), key=lambda x: int(x[1:])))
    label = ', '.join([f'{mod}({coeff:.2f})' for mod, coeff in sorted(mod_list, key=lambda x: int(x[0][1:]))])
    color = 'gray' if len(modules) > 1 else module_colors.get(modules[0], 'gray')
    dot.edge(reg, target, label=label, color=color, fontsize='13')

# === Export Graph ===
dot.render('gene_network_graph_final_0.25', view=True)
```
